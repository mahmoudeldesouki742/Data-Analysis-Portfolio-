## 2. Data Cleaning & Logical Deduction

---

### 🎯 Objective
The goal of this phase is to clean the dataset by filling missing values through mathematical deduction and business logic rules, handling invalid entries, and formatting data types properly[cite: 1, 3, 4].

---

### 💡 Logical Deduction Strategy

1. **Calculated Numerical Fields:**
   * **Missing `Total Spent`:** Calculated as $\text{Quantity} \times \text{Price Per Unit}$[cite: 2, 4].
   * **Missing `Quantity`:** Calculated as $\frac{\text{Total Spent}}{\text{Price Per Unit}}$[cite: 2, 4].
   * **Missing `Price Per Unit`:** Calculated as $\frac{\text{Total Spent}}{\text{Quantity}}$ or mapped from static item prices[cite: 1, 2, 4].

2. **Missing `Item` Inferences:**
   * Infer missing item names using `Price Per Unit` and static category pricing tables[cite: 1, 3, 4].

3. **Missing `Discount Applied` Handling:**
   * Treat missing/`None` values in `Discount Applied` as `False` (no discount applied)[cite: 2, 3].

4. **Data Type Standardization:**
   * Convert `Transaction Date` from object string to proper `datetime` format[cite: 2].

---

### Handling Missing `Item` Values

**Observation:**  
`Category + Price Per Unit` uniquely identifies the `Item`.

**Steps:**
- Created a reference table from complete rows.
- Removed duplicate combinations.
- Used the reference table to fill missing `Item` values.

**Result:**
- **200** unique `Category + Price Per Unit` combinations.
- **604** missing `Item` values successfully filled.
- **5** rows remain because `Price Per Unit` is also missing.

```python
# 1. استخراج الأعمدة الثلاثة المطلوبة
reference_table = retail_store_df[['Category', 'Item', 'Price Per Unit']].copy()

# 2. إزالة الصفوف المكررة بالكامل للحصول على قيم فريدة (Unique)
reference_table.drop_duplicates(inplace=True)

# 3. إزالة أي قيم مفرغة أو NaN إذا كانت موجودة (خطوة تنظيف فرعية)
reference_table.dropna(inplace=True)

# 4. إعادة ضبط الفهرس (Index)
reference_table.reset_index(drop=True, inplace=True)

# عرض النتيجة
reference_table
```

```python
reference_table.to_csv('reference_table.csv', index=False)
```

```python
# 1. عمل نسخة جديدة من البيانات
df_cleaned = retail_store_df.copy()

# 2. ملء القيم المفقودة في Price Per Unit عن طريق قسمة Total Spent على Quantity
df_cleaned['Price Per Unit'] = df_cleaned['Price Per Unit'].fillna(
    df_cleaned['Total Spent'] / df_cleaned['Quantity']
)

# 3. التأكد من النتيجة وعدد القيم المفقودة المتبقية
df_cleaned.isnull().sum()
```

```python
# 1. إعادة إنشاء reference_table بالقيم الفريدة للصفوف التي تحتوي على Item مكتمل
ref_table_item = df_cleaned.dropna(subset=['Category', 'Item', 'Price Per Unit'])[['Category', 'Price Per Unit', 'Item']].drop_duplicates()

# 2. عمل دمج (Merge) بناءً على Category و Price Per Unit لنسب الـ Item المطابق
df_cleaned = df_cleaned.merge(
    ref_table_item,
    on=['Category', 'Price Per Unit'],
    how='left',
    suffixes=('', '_ref')
)

# 3. ملء القيم المفقودة في Item بالقيم القادمة من المرجع
df_cleaned['Item'] = df_cleaned['Item'].fillna(df_cleaned['Item_ref'])

# 4. حذف العمود المؤقت الناتج عن الدمج
df_cleaned.drop(columns=['Item_ref'], inplace=True)

# 5. التأكد من عدد القيم المفقودة المتبقية في Item
df_cleaned.isnull().sum()
```

```python
# 1. ملء الكمية المفقودة (Quantity)
df_cleaned['Quantity'] = df_cleaned['Quantity'].fillna(
    df_cleaned['Total Spent'] / df_cleaned['Price Per Unit']
)

# 2. ملء إجمالي المبلغ المفقود (Total Spent)
df_cleaned['Total Spent'] = df_cleaned['Total Spent'].fillna(
    df_cleaned['Price Per Unit'] * df_cleaned['Quantity']
)

df_cleaned.isnull().sum()
```

```python
# ملء Quantity بالمنوال (أكثر كمية مباعة تكراراً) بحسب الـ Item
df_cleaned['Quantity'] = df_cleaned.groupby('Item')['Quantity'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 1))

# ثم حساب Total Spent
df_cleaned['Total Spent'] = df_cleaned['Total Spent'].fillna(
    df_cleaned['Price Per Unit'] * df_cleaned['Quantity']
)
df_cleaned.isnull().sum()
```

```python
# 1. حساب السعر الأصلي المتوقع بدون خصم
expected_total = df_cleaned['Price Per Unit'] * df_cleaned['Quantity']

# 2. استنتاج قيمة الخصم بناءً على الفرق بين Total Spent و expected_total
# إذا كان هناك فرق (Total Spent أقل من Expected Total)، إذن تم تطبيق خصم (True)، وغير ذلك (False)
calculated_discount = df_cleaned['Total Spent'] < expected_total

# 3. ملء القيم المفقودة (null/NaN) بالنتيجة المحسوبة دقيقاً
df_cleaned['Discount Applied'] = df_cleaned['Discount Applied'].fillna(calculated_discount)

# التأكد من عدم وجود أي قيمة مفقودة
print("عدد القيم المفقودة المتبقية في Discount Applied:", df_cleaned['Discount Applied'].isna().sum())
print("\nتوزيع القيم الحقيقي بعد الاستنتاج:")
print(df_cleaned['Discount Applied'].value_counts())
```
